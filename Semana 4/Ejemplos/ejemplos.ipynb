```python
#Programa operaciones mutables en listas
# "del" es una instrucción no un método.
# El método sort() no funciona cuando se trata de una lista con elementos numéricos y alfabéticos.

colores=["Rojo","Azul","Gris","Rojo","Amarillo"]
print("Lista original: ",colores)
del colores[2]
print("Lista eliminando la posición 2: ",colores)

colores.append("Gris")
print("Lista anexando el color Gris: ",colores)

colores.insert(1,"Blanco")
print("Lista intercalando el color Blanco: ",colores)

colores.remove("Rojo")
print("Lista quitando la primera aparición de Rojo: ",colores)

colores.reverse()
print("Lista con orden invertido: ",colores)

colores.sort()
print("Lista ordenada ascendentemente: ",colores)

colores.sort(reverse=True)
print("Lista ordenada descendentemente: ",colores)
```

```python
#Programa operaciones inmutables en listas
#  "in" consiste en una instrucción y no precisamente en un método
# 

calGpo1=[79,94,79,97,77]
calGpo2=[82,72,100,96]

print("Calificaciones del grupo 1: ",calGpo1,"\nCalificaciones del grupo 2: ",calGpo2)
print("Alumnos del grupo 1: ",len(calGpo1),"\nAlumnos del grupo 2: ",len(calGpo2))
print("Calificaciones de ambos grupos: ",calGpo1+calGpo2)
print("Calificaciones del grupo 2 replicadas 3 veces: ", calGpo2*3)
print("Calificaciones más bajas de cada grupo: ",min(calGpo1)," y ",min(calGpo2))
print("Calificaciones más altas de cada grupo: ",max(calGpo1)," y ",max(calGpo2))
print("Alumnos con calificación de 79 en cada grupo: ",calGpo1.count(79)," y ",calGpo2.count(79))
print("Suma de Calificaciones de cada grupo: ",sum(calGpo1)," y ",sum(calGpo2))
print("¿hubo calificación perfecta en el grupo?: ",100 in calGpo1," y ",100 in calGpo2)
print("Calificaciones del grupo 1: ",calGpo1,"\nCalificaciones del grupo 2: ",calGpo2)
```

```python
#Programa lista de múltiplos de 3 del 0 al 1,000
mult3=[]

for i in range(1000):
    if i % 3 == 0:
        mult3.append(i)
print(mult3)
print("\nTotal de elementos: ",len(mult3))
```

```python
#Programa "Capturar nombres de personas"
personas=[]

while True:
    nombre=input("Dame el nombre de la persona, enter para terminar: ")
    if nombre!="":
        personas.append(nombre)
    else:
        break
print("\n",personas)
personas.sort()
print("Lista ordenada ascendentemente: ",personas)
```
