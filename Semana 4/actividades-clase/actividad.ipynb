````python
def promedio(sumatoria, n):
    return sumatoria / n

n = int(input("¿Cu´ántas calificaciones quiere promediar?"))
sumatoria = 0

print(f"Se van a pedir {n} calificaciones")
for c in range (1, n+1):
    calif = float(input(f"Dame la calificación {c}"))
    sumatoria = sumatoria + calif
   # print(f"Hasta este punto la sumaotria va en {sumatoria}")
print(f"El promedio de {n} calificaciones es {promedio(sumatoria, n)}")
````
