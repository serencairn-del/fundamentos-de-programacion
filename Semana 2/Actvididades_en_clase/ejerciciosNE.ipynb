**Ejercicio 4: Clasificación por rango de edad con elif** 
* Enunciado: Crea un programa que clasifique a una persona según su edad: menor de 3 años "Bebé", de 3 a 12 "Niño", de 13 a 17 "Adolescente" y de 18 en adelante "Adulto". Usa elif. Entrada: 15 Salida: Adolescente 

```python
rango = int(input("Ingresa la edad: "))

if( rango >=0 and rango <3):
    print("Eres bebé")
elif(rango >=3 and rango <12):
    print("Eres niño")
elif(rango >=12 and rango <17):
    print("Eres adolescente")
else:
    print("Eres adulto")

    

```

**Ejercicio 5: Descuento port tipo de visitante**
* Enunciado: Crea un programa que pida el tipo de visitante (adulto_mayor, profesor, estudiante u otro) y calcule el precio de un boleto base de $45 aplicando solo un descuento: adulto mayor 12%, profesor 10%, estudiante 10%, otro 0%. Entrada: estudiante Salida: Precio final: $40.50

```python
tipo = str("¿Qué eres? (adulto_mayor, profesor, estudiante, otro)")

precio = 45 

if tipo == "adulto_mayor":
    descuento = precio * 0.12    
elif tipo == "profesor" or tipo == "estudiante":
    descuento = precio * 0.10       
else:
    descuento = 0
total = precio - descuento
print(f"Total a pagar: ${total:.2f}") 


```
