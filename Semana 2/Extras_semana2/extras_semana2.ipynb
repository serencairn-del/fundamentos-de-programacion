### Extra 1: Control de aforo con break y continue (while)
**Enunciado:** Crea un programa que procese visitantes del museo con un ciclo `while`. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa `continue`). Si el total acumulado alcanza o supera $100, detén el ciclo con `break` y muestra el total.
**Entrada:**
```
Costo del boleto 1: 45
Costo del boleto 2: 0
Costo del boleto 3: 30
Costo del boleto 4: 45
```
**Salida:**
```
Menor de 3 años, sin cargo
Aforo alcanzado, se detiene el registro
Total acumulado: $120.00

**Modelo EPS**

* ENTRADA:
    * Capturar el número de visitantes y determinar el precio que pagarán dependiendo su edad.
    * Mostrar cuando el boleto sea gratis (menores de 3 años) y usar 'continue' para pasar al siguiente visitante.
    * Si el monto acumulado por las personas supera los $100 usar 'break' y mostrar el total.

````python
totalfinal = 0

while True:
    costo = int(input("¿Cuál es el costo de tu boleto?"))
    

    if costo == 0:
        print("Menor de 3 años, sin cargo.")
        continue
    totalfinal = totalfinal + costo

    if totalfinal >=100:
        print("Aforo alcanzado, se detiene el registro.")
        break

print(f"Total acumulado: ${totalfinal}.00")

````

### Extra 2: Estadística de visitantes con for
**Enunciado:** Crea un programa que pida el número total de visitantes y, con un ciclo `for`, capture la edad de cada uno. Debe contar cuántos son adultos (18 años o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos resultados con dos decimales.
**Entrada:**
```
Número de visitantes: 4
Edad del visitante 1: 30
Edad del visitante 2: 45
Edad del visitante 3: 15
Edad del visitante 4: 2
```
**Salida:**
```
Adultos: 2
Promedio de edad: 23.00
```

***Modelo EPS***
*ENTRADA
    * Pedir el número entero de visitantes y guardar ese número en una variable.
*PROCESO
    * El ciclo for repite el código dependiendo de los visitantes se definan mientras que la condicional evalua si la edad de la persona se cumple, si se cumple se guarda en el contador. Fuera del ciclo se divide la edad acumulada entre el total de las personas.
*SALIDA
    * Mostrar los resultados al final

````python
personas = int(input("¿Cuántas personas son? "))
edad_acumulada = 0
contador_adultos = 0


for p in range(1, personas + 1):
    edad_persona = int(input(f"Dame la edad de la persona {p}"))
    if edad_persona >= 18:
        contador_adultos = contador_adultos + 1
    edad_acumulada = edad_acumulada + edad_persona
    print(f"Cantidad de visitantes: {personas}")
    print(f"Edad del visitante {p}: {edad_persona}")

promedio = edad_acumulada / personas
print (f"Los adultos totales son: {contador_adultos}")
print(f"El promedio de edad es: {promedio:.2f}")
````

### Extra 3: Depuración de un cobro con PDB
**Enunciado:** El siguiente código pretende calcular el precio final de un boleto de $45 con el descuento del adulto mayor (12%) pero contiene un error de lógica. Inserta `pdb.set_trace()` para rastrear las variables, identifica el error y escribe la versión corregida.
```python
precio = 45
descuento = 12
total = precio - descuento
print(f"Total: ${total:.2f}")
```
**Entrada:** Ninguna (código dado para depurar).
**Salida:**
```
Error de lógica: descuento debe ser el porcentaje 0.12, no 12.
Código corregido: descuento = precio * 0.12 -> Total: $39.60

````python
precio = 45
descuento = 12
total = precio - descuento
print(f"Total: ${total:.2f}")

# Código corregido
precio = 45 
descuento = (45 * 0.12)
total = precio - descuento
print(f"Total correcto: ${total:.2f}")
````

### Extra 4: Pirámide de asteriscos con for
**Enunciado:** Crea un programa que pida la altura de una pirámide y la dibuje con asteriscos usando ciclos `for` anidados y el parámetro `end` de `print`.
**Entrada:**
```
Altura de la pirámide: 4
```
**Salida:**
```
*
**
***
****
```

***Modelo EPS***
* Entrada: el usuario tiene que poner la altura de la piramide; Proceso: el primer ciclo for se encarga de registrar la altura = filas verticales y el segundo ciclo for se encarga de aumentar los asteriscos a cada columna agregando un asterisco más. El end="" evita que python salte la linea automaticamente; Salida: mostrar la piramide de asteriscos.

````python
alturap = int(input("¿Cuál será a altura de tu piramide?: "))
for e in range(alturap):
    for s in range ( e + 1):
        print("*", end="")
    print()
````
