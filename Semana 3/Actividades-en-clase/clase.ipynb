### Ejercicio 2: Años bisiestos en un rango
**Enunciado:** Crea un programa que pida dos años (inicio y fin) y muestre en una línea todos los años bisiestos de ese rango. Regla: un año es bisiesto si es divisible entre 4 pero no entre 100, o si es divisible entre 400. Usa el `for` con `range` y operadores lógicos (`and`/`or`).
**Entrada:** `2000` y `2010`
**Proceso:** `2000 % 400 == 0` (bisiesto), `2004 % 4 == 0` y `2004 % 100 != 0` (bisiesto), `2008` (bisiesto)
**Salida:** `Años bisiestos: 2000, 2004, 2008`

````python
a_inicio = int(input("Ingresa año: "))
a_final = int(input("¿Cuál es el ultimo año? "))

print("Años bisiestos:", end=" ")

for anio in range (a_inicio, a_final + 1):
    if(anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
        print(anio, end="\t")
````

### Ejercicio 7: Desglose de billetes
**Enunciado:** Un cajero entrega el monto solicitado con la menor cantidad de billetes posibles. Crea un programa que pida un monto (múltiplo de 10) y muestre cuántos billetes de `$500`, `$200`, `$100`, `$50`, `$20` y monedas de `$10` entregará, usando `//` y `%`.
**Entrada:** `1380`
**Proceso:** `1380 // 500 = 2` (restan 380), `380 // 200 = 1` (restan 180), y así sucesivamente
**Salida:**
```
Billetes de $500: 2
Billetes de $200: 1
Billetes de $100: 1
Billetes de $50: 1
Billetes de $20: 1
Monedas de $10: 1

***Explicación:*** input() siempre devuelve lo que el usuario escribe como texto (string/str) aunque escriba números; por eso se envuelve con int() ya que esto convierte ese texto a número entero para poder hacer operaciones matemáticas con este, al no hacer la conversión no podríamos usar // (operador de división entera) y % (operador modulo). La línea "billetes500 = monto // 500" y ñas siguientes que siguen hacen la división entera; divide dos números y se queda solo con la parte entera del resultado, sin decimales. Esto nos dice cuantos billetes de 500 caben en el monto dado por el usario. La línea debajo: "monto = monto % 500" nos da el residuo, lo que sobra en la divisón entera; en esta linea actualizamos la variable de monto, el monto original se queda olvidado y monto ahora vale lo que queda por repartir (residuo). Cada bloque trabaja trabaja sobre lo que sobró del bloque anterior. Entonces los siguientes de billetes hacen lo mismo, pero con la cantidad que va quedando cada vez y así con el operador de divisón entera se calcula cuantos billetes de esa denominación caben y se actualiza monto con lo que sobra utilizando el operador modulo. En "monedas10" ya no requerimos el operador de modulo ya que el monto al ser multiplo de 10 no debería sobrar nada. Con los print, mostramos cuantos billetes de cada denominación y se darían.

Lo importante es tener un orden, de mayor a menor para ir "vaciando" ya que cada vez que sacas billetes de una denominación te quedas con residuo y ese residuo pasa al siguiente bloque para seguir repartiendo; así usamos la menor cantidad de billetes en total; si esto no se hicera el resultado podría ser que nos de más billetes de los necesarios. ¿Por qué dividir entre 500, 200, 100, 50, 20 y 10? más que nada por las denominaciones dadas.

````python
# Vamos a pedir al usuario el dato (monto) al usuario y tenemos que definir que tipo de dato es, en este caso será entero.
monto = int(input("Ingresa el monto (multiplo de 10): "))

billetes500 = monto // 500
monto = monto % 500

billetes200 = monto // 200
monto = monto % 200

billetes100 = monto // 100
monto = monto % 100

billetes50 = monto // 50
monto = monto % 50

billetes20 = monto // 20
monto = monto % 20

monedas10 = monto // 10

print("Billetes de $500:", billetes500)
print("Billetes de $200:", billetes200)
print("Billetes de $100: ", billetes100)
print("Billetes de $50: ", billetes50)
print("Billetes de $20: ", billetes20)
print("Mondeas de $10: ", monedas10)
````
