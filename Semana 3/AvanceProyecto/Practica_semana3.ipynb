```python
# EJERCICIO 2

a_inicio = int(input("Ingresa año: "))
a_final = int(input("¿Cuál es el ultimo año? "))

print("Años bisiestos:", end=" ")

for anio in range (a_inicio, a_final + 1):
    if(anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
        print(anio, end="\t")
```
