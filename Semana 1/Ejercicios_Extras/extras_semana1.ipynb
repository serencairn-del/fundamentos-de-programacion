### Ejercicio extra 1

*   **Entrada:**
    * Total_Cuenta (Es un numero flotante ya que puede ser inexacto)
    * porcentaje_p (También es un numero flotante ya que puede ser inexacto)
    * n_personas (Número entero)
*   Proceso:
    * Propina = Total_Cuenta * (porcentaje / 100) (Sacamos el porcentaje del 15% de la cuenta y dividimos entre 100)
    * TTLconPropina = Total_Cuenta + propina (sumamos la propina con el total de la cuenta)
    * pagoxpersona = TTLconPropina / personas (dividimos el pago con la propina ya incluida entre el nuúmero de personas presentes)
*   Salida:
    * Muestra el monto de la propina, la cuenta total incluyendo la propina y el resultado de la división

```python
Total_Cuenta = float(input("El total de la cuenta es: "))
porcentaje_p = float(input("El procentaje de la propina es de: "))
n_personas = int(input("¿Cuál es el número de personas?"))

propina = Total_Cuenta * (porcentaje_p / 100)
TTLconPropina = Total_Cuenta + propina
pagoxpersona = TTLconPropina / n_personas

print(f"La propina es de: ${propina:.2f}")
print(f"Monto a pagar con propina incluida: ${TTLconPropina:.2f}")
print(F"El pago por persona es: ${pagoxpersona:.2f}")
```

### Ejercico extra 2
**Entrada:**
* 'minutos' (Aquí se ingresa un total de minutos, siendo número entero/exacto)
**Proceso:**
* 'd = minutos // 1440' (se divide la cantidad de minutos entre el total de mimnutos de un día para saber si cuantos días hay)
* 'm_rest = minutos % 1440' (el simnbolo % sirve para saber si nos sobraron minutos en la división y que no formaoron un día completo)
* 'h = m_rest // 60 (aquí podemos identificar cuantas horas hay en lo que sobró anteriormente)
* min_f = m_rest % 60 (volvemos a repetir el % para saber si hay minutos que no alcanzaron a formar una hora)
**Salida:** 
* usamos print para mostrar los días, horas y minutos; el + no suma sino que junta (concatena) el texto.

```python
minutos = int(input("Ingresa una cantidad de minutos: "))

d = minutos // 1440 
m_rest = minutos % 1440
h = m_rest // 60 
min_f = m_rest % 60 

print(f" {minutos} minutos + {d} día(s) + {h} hora(s) + {min_f} minutos")
```

### Ejercicio extra 3
* ENTRADA (definimos los tres parciales teniendo en cuenta que la calificación es de 0 a 10)
    * 'parcial'
    * 'parcial2'
    * 'parcial3'
* PROCESO (hay que asignar el porcentaje para cada parcial 30%, 30%, 40%)
    * 'notafinal = (parcial1 * 0,30) + (parcial2 * 0.30) + (parcial3 * 0.40)' (aqui usamos el metodo decimal para hacer conversión de porcentajes por eso 30 pasa a ser 0.30; multiplicamos (variable * porcentaje) sumamos (usamos + entre cada parentesis) y así el total de los 3 parciales
* SALIDA (usamos print para mostrar la calificación final y tenemos que mostrarlo con dos decimales usando :.2f)

```python


parcial1 = float(input("Ingresa calificación PARCIAL 1:: "))
parcial2 = float(input("Ingresa calificación PARCIAL 2: "))
parcial3 = float(input("Ingresa calificación PARCIAL 3: "))

notafinal =  (parcial1 * 0.30) + (parcial2 * 0.30) + (parcial3 * 0.40)

print(f"La calificación final es: {notafinal:.2f}")
```

### Ejercicio extra 4
*extranjera a nacional o si se quiere saber cuántas unidades extranjeras dan por el dinero, dividir la cantidad entre el tipo de cambio*


* ENTRADA
    * 'cpesos' (monto en pesos mexicanos)
    * 'dolar' (valor del dolar en pesos) El propio usuario debe saber el valor de USD.
    * 'euros' (valor del euro en pesos) El propio usuario deber saber el valor de EUR.
* PROCESO
    * 'dolars = pesos / * dolar'
    * 'euross = pesos / * euros'
* SALIDA (USAMOS PRINT PARA MOSTRAR LA CANTIDAD INCIAL EN PESOS MEXICANOS , LUEGO A DOLARES Y POR ULTIMO A EUROS JUNTO A DOS DECIMALES)


```python
cpesos = float(input("Ingresa la cantidad en pesos MXN: "))
dolar = float(input("Ingresa el valor de USD: "))
euros = float(input("Ingresa el valor de EUR"))

dolars = cpesos / dolar
euross = cpesos / euros

print(f"El monto en MXN: {cpesos}")
print(f"Conversión USD: {dolars:.2f}")
print(f"Conversión EUR: {euross:.2f}")
```
