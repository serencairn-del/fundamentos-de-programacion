# CALCULADORA

```python
usuario = input("Ingresa tu nombre de usuario: ")

print ( f"Hola {usuario}.") 

redes_ig = float(input("¿Cuántas horas pasas en Instagram?: "))
mensajeria_Teg = float(input("¿Cuántas horas usas Telegram?: "))
streamg_netflix = float(input("¿Cuántas horas te sientas a ver Netflix?"))
games_steam = float(input("¿Cuántas horas juegas gracias a Steam?: "))
estudio_canvas = float(input("¿Cuántas horas dedicas a tu estudio en Canvas?"))

sumaTmp = redes_ig + mensajeria_Teg + streamg_netflix + games_steam + estudio_canvas

porcentaje = (sumaTmp / 24) * 100


print(f"El total de horas que gastas en estás plataformas es de: {sumaTmp}")
print(f"Este el porcentaje de tu día: {porcentaje:.2f}")
```
