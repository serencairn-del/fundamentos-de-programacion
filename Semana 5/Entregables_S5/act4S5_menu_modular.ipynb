```python
numeros = (9, 3, 102, 39, 99, 24, 12, 21, 84)

contactos = {"REM": "555-999", "RYU": "555-333", "LUZ": "555-111"}

mensaje = "Python es un lenguaje poderoso"

def sumar_tupla(tupla_datos):
    total = 0
    for num in tupla_datos:
        total += num
    return total

def modulodeTuplas():
    print("=== SECCIÓN: TUPLAS ===")
    print(f"El tercer elemento de la tupla original es: {numeros[2]}")

    n1 = int(input("Ingresa un nuevo número: "))
    n2 = int(input("Ingresa otro número: "))
    nuevatupla = numeros + (n1,n2)

    listaT = list(nuevatupla)
    listaT.sort()

    print("Lista ordenada: ")
    for e in listaT:
        print(e, end=", ")
    print()

    resultadosuma = sumar_tupla(nuevatupla)
    print(f"La suma de los elementos es: {resultadosuma}")


def buscaragenda(diccionario, nombre):
    return diccionario.get(nombre, "No encontrado c:")

def modulodediccionarios():
    print("=== SECCIÓN: DICCIONARIOS === ")

    nuevonombre = input("Nombre del nuevo contacto: ")
    nuevotelefono = input("Telefono del nuevo contacto: ")
    contactos[nuevonombre] = nuevotelefono

    print("Nombres de contactos registrados: ", end="")
    for nombre in contactos.keys():
        print(nombre, end=", ")
    print()

    buscarnombre = input("Ingresa el nombre del contacto que quieres buscar: ")
    telobt = buscaragenda(contactos, buscarnombre)
    print(f"El telefono de {buscarnombre} es: {telobt}")


def modulodeexcepciones():
    print("=== SECCIÓN: EXCEPCIONES ===")
    try:
        n1 = int(input("Ingresa el primer número: "))
        n2 = int(input("Ingresa el segundo número: "))
        resultado = n1/n2
        print(f"Primer número: {n1}")
        print(f"Segundo número: {n2}")
        print(f"El resultado es: {resultado}")
    except ZeroDivisionError:
        print("No se puede dividir entre cero :)")
    except ValueError:
        print("Solo números enteros.")


def contarpalabras(txt):
    return len(txt.split())

def mdl_str():
    print("=== SECCIÓN: CONTEO (STRINGS) ===")
    print(f"Mensaje original: {mensaje}")
    print(f"Longitud del mensaje: {len(mensaje)}")
    print(f"En mayúsculas: {mensaje.upper()}")

    mensaje_nuevo = mensaje.replace("Python", "Programación")
    print(f"Texto reemplazado: {mensaje_nuevo}")

    totalpalabras = contarpalabras(mensaje)
    print(f"Cantidad de palabras: {totalpalabras}")

opcion = 0

while opcion != 5:
    print("=== MENÚ PRINCIPAL ===")
    print("1. Sección de Tuplas")
    print("2. Sección de Diccionarios")
    print("3. Sección de Excepciones")
    print("4. Sección de Strings")
    print("5. Finalizar programa")

    try:
        opcion = int(input("Elige una opción del 1 al 5: "))
        print()

        if opcion == 1:
            modulodeTuplas()
        elif opcion == 2:
            modulodediccionarios()
        elif opcion == 3:
            modulodeexcepciones()
        elif opcion == 4:
            mdl_str()
        elif opcion == 5:
            print("Saliendo del programa c:")
        else:
            print("Opción no válida. Selecciona solo del 1 al 5 ^^")
    except ValueError:
        print("Entrada incorrecta. Debes teclear un número entero del 1 al 5 c:")
```
