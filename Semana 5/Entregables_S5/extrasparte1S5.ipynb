```python
# EJERCICIO 1

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

nuevanota1 = float(input("Nueva calificación 1: "))
nuevanota2 = float(input("Nueva calificación 2: "))

nuevatupla = calificaciones + (nuevanota1, nuevanota2)

listaTpla = list(nuevatupla)
listaTpla.sort(reverse=True)

def sumaTupla(t):
    total = 0
    for n in t:
        total = total + n
    return total

sumatotal = sumaTupla(nuevatupla)

print("Tercera calificación:", (calificaciones[-3]))
print(f"Nueva tupla: {nuevatupla}")
print(f"Lista ordenada (mayor a menor): {listaTpla}")
print(f"Suma total: {sumatotal}")
```

```python
#EJERCICIO 2

agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

nuevoN = input("Nombre del nuevo contacto: ")
nuevoC = input("Celular del nuevo contacto: ")

agenda[nuevoN] = nuevoC

buscar = input("Nombre a buscar: ")

print("CONTACTOS REGISTRADOS: ", end="")
for nom in agenda.keys():
    print(nom, end=", ")
print()

def buscar_telefono(nommbrecontacto):
    return agenda.get(nommbrecontacto, "No encontrado")

resultado = buscar_telefono(buscar)
print(f"El telefono de {buscar} es: {resultado}")
```

```python
# EJERCICIO 3

try: 
    primerE = int(input("PRIMER NÚMERO ENTERO: "))
    segundaE = int(input("SEGUNDO NÚMERO ENTERO: "))
    div = primerE / segundaE
    print(f"El resultado de la división de {primerE} entre {segundaE} es: {div}")
except ValueError:
    print("SOLO DEBES INGRESAR NÚMEROS ENTEROS")
except ZeroDivisionError:
    print("NO ES POSIBLE DIVIDIR ENTRE CERO ^^. INTENTA DE NUEVO.")
```

```python
#EJERCICIO 4

mensaje = "Python es un lenguaje poderoso"

def conteo(tex):
    return len(tex.split())

total = conteo(mensaje)

print("Longitud del mensaje: ", len(mensaje))
print("En mayúsculas: ", mensaje.upper())
print(mensaje.replace("Python", "Programación"))
print("Palabras totales: ", total)
```
