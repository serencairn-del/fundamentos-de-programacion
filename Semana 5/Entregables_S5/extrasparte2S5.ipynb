```python
# DESAFIO EXTRA 2

candidatos = {
    "E": 0,
    "X": 0,
    "C": 0 
}

cantidad = int(input("Cuántos votos van a ser: "))
conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A, B o C): ")
    if voto.upper() in candidatos:
        print(f"Voto válido para {voto.upper()}")
        candidatos[voto.upper()] +1
        conteo += 1
    else:
        print("El voto no es válido, vota nuevamente")

print("\nResultados: ")
for k,v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
```

```python
# DESAFIO EXTRA 3


while True:
    try: 
        numero1 = int(input("Ingresa el primer valor: "))
        numero2 = int(input("Ingresa el segundo valor: "))

        break
    except ValueError:
        print("DEBES INGRESAR UN NUÚERO ENTERO VÁLIDO. INTENTA DE NUEVO")

try: 
    resultado = numero1/numero2
    print(f"Los números de la división son: {numero1}/{numero2}")
    print(f"El resultado de la división: {resultado}")

except ZeroDivisionError:
    print("NO SE PUEDE DIVIVIDR ENTRE CERO ^^")
```
