```python
# DESAFIO EXTRA 2

candidatos = {
    "E": 0,
    "X": 0,
    "C": 0 
}

cantidad = int(input("Cuántos votos van a ser: "))
conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A, B o C): ")
    if voto.upper() in candidatos:
        print(f"Voto válido para {voto.upper()}")
        candidatos[voto.upper()] +1
        conteo += 1
    else:
        print("El voto no es válido, vota nuevamente")

print("\nResultados: ")
for k,v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
```

### Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario `contactos`, (3) dividir dos números con manejo de `ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
# DESAFÍO EXTRA 5

agenda = {"Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"}
numerostpl = (9,3,12,99,102,39,54,66,33)

def opcion_t():
    print("ORDENAR TUPLA ")
    lista = list(numerostpl)
    lista.sort()
    print(lista)

def opcionagenda():
    print("BUSCAR TELEFONO ")
    print(agenda.keys())
    buscarnom = input("Ingresa el nombre del contacto: ")
    telefono = agenda.get(buscarnom, "No encontrado")
    print(f"El telefono de {buscarnom} es: {telefono}")

def division():
    try:
        n1 = int(input("Ingresa el primer número: "))
        n2 = int(input("Ingresa el segundo número: "))
        resultado = n1/n2
        print(f"Primer número: {n1}")
        print(f"Segundo número: {n2}")
        print(f"El resultado es: {resultado}")
    except ZeroDivisionError:
        print("No se puede dividir entre cero :)")
    except ValueError:
        print("Solo números enteros.")

def analizartxt():
    print("Analizar longitud, mayúsculas y cantidad de palabras")
    txt = input("Introduce un mensaje: ")
    print(f"Longitud: {len(txt)}")
    print(f"Mayúsculas: {txt.upper()}")
    print(f"Palabras totales: {len(txt.split())}")

# MENU

opcion = 0

while opcion != 5:
    print("\n|||MINIMENÚ INTEGRADOR |||")
    print("1. Mostrar tupla ordenada")
    print("2. Buscar teléfono en la agenda")
    print("3. Dividir dos números")
    print("4. Analizar un mensaje de texto")
    print("5. Salir")

    try:
        opcion = int(input("Elige una opción del 1 al 5"))
        print()

        if opcion == 1:
            opcion_t()
        elif opcion == 2:
            opcionagenda()
        elif opcion == 3:
            division()
        elif opcion == 4:
            analizartxt()
        elif opcion == 5:
            print("Finalizando programa c:")
        else:
            print("Opción no válida, intenta de nuevo.")
    except ValueError:
        print("\nPor favor ingresa solo números del 1 al 5.")
```
